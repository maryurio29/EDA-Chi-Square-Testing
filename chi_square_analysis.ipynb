```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from scipy.stats.contingency import association

# Load the data from churn_clean.csv
df = pd.read_csv("churn_clean.csv")

# Check for number of observations and variables
df.shape

# Check for missing values by column
empty_counts = df.isna().sum()
print(empty_counts)

# 'None' is used in the 'InternetService' column
df['InternetService'] = df['InternetService'].fillna('No_Internet')

# Check empty_counts
empty_counts = df.isna().sum()
print(empty_counts)

# Counts for Churn categories
churn_value_counts = df['Churn'].value_counts()
print(churn_value_counts)

# Counts for Marital categories
marital_value_counts = df['Marital'].value_counts()
print(marital_value_counts)

# Contingency Table for Churn and Marital status
contingency_table_churn_marital = pd.crosstab(df.Churn, df.Marital)
print(contingency_table_churn_marital)

# Chi-Square Test for Churn and Marital status
marital_chi2_test = chi2_contingency(contingency_table_churn_marital)
print('Chi-Square Value = ', marital_chi2_test.statistic)
print('P Value = ', marital_chi2_test.pvalue)
print('Degrees of Freedom = ', marital_chi2_test.dof)
if (marital_chi2_test.pvalue > 0.05):
    print('P value is greater than 0.05 therefore we fail to reject the null hypothesis and accept it')
else:
    print('P value is less than 0.05 therefore we reject the null hypothesis and accept the alternative hypothesis')
```

```python
# Counts for InternetService categories
internetService_value_counts = df['InternetService'].value_counts()
print(internetService_value_counts)

# Contingency Table for Churn and InternetService
contingency_table_churn_internet = pd.crosstab(df.Churn, df.InternetService)
print(contingency_table_churn_internet)

# Chi-Square Test for Churn and InternetService status
internet_chi2_test = chi2_contingency(contingency_table_churn_internet)
print('Chi-Square Value = ', internet_chi2_test.statistic)
print('P Value = ', internet_chi2_test.pvalue)
print('Degrees of Freedom = ', internet_chi2_test.dof)
if (internet_chi2_test.pvalue > 0.05):
    print('P value is greater than 0.05 therefore we fail to reject the null hypothesis and accept it')
else:
    print('P value is less than 0.05 therefore we reject the null hypothesis and accept the alternative hypothesis')

# Pearson's correlation coefficient for Tenure and Outage_sec_perweek variables:
pearson_cc = df['Tenure'].corr(df['Outage_sec_perweek'])
print("Pearson's correlation coefficient = ", pearson_cc)

# Cramer's V calculation for Churn and InternetService variables
cramer_v = association(contingency_table_churn_internet, method="cramer")
print("Cramer's V = ", cramer_v)
```
